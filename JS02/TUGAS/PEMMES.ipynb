```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler, LabelEncoder
```

```python
df = pd.read_csv('wbc.csv', sep=';')
```

#Tugas 1 Pemisahan Variabel yang dapat dan tidak dapat digunakan

```python
print("Dataset awal:")
print(df.head())
print("\nInformasi Dataset:")
df.info()

df_clean = df.drop(['id'], axis=1)

print("\n=== HASIL PEMISAHAN VARIABEL ===")
print(f"Jumlah variabel yang TIDAK digunakan: 1 (kolom 'id')")
print(f"Jumlah variabel yang DAPAT digunakan (fitur): {df_clean.shape[1] - 1} (semua kolom kecuali 'diagnosis')")
print(f"Jumlah variabel target: 1 (kolom 'diagnosis')")
print("\nDaftar variabel yang DAPAT digunakan:")
print(df_clean.columns.tolist())
```

#Tugas 2 Melakukan proses encoding pada kolom "diagnosis"

```python
df_clean = df.drop(['id'], axis=1)

le = LabelEncoder()
df_clean['diagnosis'] = le.fit_transform(df_clean['diagnosis'])

print("\n=== HASIL ENCODING KOLOM DIAGNOSIS ===")
print(f"Kelas asli: {le.classes_}")  # ['B', 'M']
print(f"Encoding: B -> {le.transform(['B'])[0]}, M -> {le.transform(['M'])[0]}")
print("\n5 data pertama setelah encoding:")
print(df_clean.head())
print("\nDistribusi hasil encoding:")
print(df_clean['diagnosis'].value_counts())
print(f"  {df_clean['diagnosis'].value_counts()[0]} sampel Benign (0)")
print(f"  {df_clean['diagnosis'].value_counts()[1]} sampel Malignant (1)")
```

#Tugas 3 Melakukan proses standardisasi pada semua kolom yang memiliki nilai numerik


```python
df_clean = df.drop(['id'], axis=1)

le = LabelEncoder()
df_clean['diagnosis'] = le.fit_transform(df_clean['diagnosis'])

X = df_clean.drop(['diagnosis'], axis=1)
y = df_clean['diagnosis']

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

X_scaled_df = pd.DataFrame(X_scaled, columns=X.columns)

df_final = X_scaled_df.copy()
df_final['diagnosis'] = y.values

print("\n=== HASIL STANDARDISASI ===")
print("Statistik data setelah standardisasi (seharusnya mean≈0, std≈1):")
print("\nMean (seharusnya mendekati 0):")
print(X_scaled_df.mean().round(10))
print("\nStd (seharusnya mendekati 1):")
print(X_scaled_df.std().round(10))

print("\n5 data pertama setelah standardisasi:")
print(df_final.head())

print(f"\nJumlah fitur setelah standardisasi: {X_scaled_df.shape[1]}")
print(f"Jumlah sampel: {X_scaled_df.shape[0]}")

df_final.to_csv('wbc_final_processed.csv', index=False)
print("\nData hasil preprocessing telah disimpan ke 'wbc_final_processed.csv'")
```
