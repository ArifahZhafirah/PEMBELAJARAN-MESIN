## PENJELASAN LIBRARY

---

**PyPREP** adalah library yang digunakan untuk membersihkan data EEG atau rekaman sinyal otak. Library ini berfungsi mendeteksi saluran yang rusak, menghilangkan interferensi listrik, memperbaiki referensi sinyal, dan menginterpolasi channel yang bermasalah sehingga menghasilkan data EEG yang bersih dan siap dianalisis lebih lanjut untuk keperluan penelitian neurosains atau pengembangan Brain-Computer Interface.

**Scipy** merupakan library fundamental untuk komputasi saintifik di Python yang menyediakan berbagai modul matematika dan teknis. Modul-modul utamanya mencakup pemrosesan sinyal untuk filtering dan deteksi puncak, analisis frekuensi dengan transformasi Fourier, statistika untuk uji hipotesis, optimasi fungsi, integrasi numerik, dan aljabar linear. Library ini sangat penting dalam pemrosesan sinyal biomedis seperti filtering ECG, deteksi QRS, dan analisis spektral.

**Wandb** atau Weights & Biases adalah platform untuk melacak dan memvisualisasikan eksperimen machine learning. Library ini secara otomatis mencatat hyperparameter dan metrik performa selama proses training, menampilkannya dalam dashboard interaktif berbasis web, serta menyimpan artifact seperti dataset dan model. Wandb membantu peneliti menjaga reproduktifitas penelitian dan memudahkan perbandingan antar eksperimen.

**PyECG** adalah toolkit khusus untuk membaca dan menganalisis data ECG atau rekaman jantung. Library ini mampu membaca file ECG dalam format WFDB dan iShine Holter, mendeteksi kompleks QRS untuk menghitung denyut jantung, serta menyediakan fungsi filtering dasar. PyECG sangat berguna bagi mahasiswa atau peneliti yang bekerja dengan data jantung untuk keperluan analisis irama jantung dan penelitian kardiologi.

---

Secara keseluruhan, keempat library ini saling melengkapi dalam mendukung penelitian sinyal biomedis dan machine learning. PyPREP dan pyECG berperan dalam pengolahan data awal untuk sinyal otak dan jantung, Scipy menjadi fondasi perhitungan saintifik dan pemrosesan sinyal, sedangkan wandb berfungsi sebagai sistem pelacakan eksperimen. Dengan menguasai keempatnya, mahasiswa telah memiliki perangkat lengkap untuk melakukan penelitian dari tahap pembersihan data hingga pembuatan model machine learning yang terdokumentasi dengan baik.

#1. Contoh Pelanggaran Etika dan Hukum dalam Penggunaan Kecerdasan Buatan

Salah satu contoh paling menonjol adalah penyalahgunaan teknologi AI generatif untuk membuat gambar pornografi palsu tanpa persetujuan subjeknya, seperti kasus gambar Taylor Swift yang beredar luas di media sosial. Dari kasus ini muncul tiga persoalan hukum sekaligus: pelanggaran privasi karena wajah seseorang digunakan tanpa izin untuk konten yang tidak senonoh, ketiadaan pembatasan hukum yang jelas terhadap penyebaran konten semacam ini di masyarakat, sekaligus dorongan bagi aparat penegak hukum untuk merumuskan cara menindak penyalahgunaan AI-generated content. Di Indonesia, kasus semacam ini bersinggungan dengan Undang-Undang Nomor 27 Tahun 2022 tentang Pelindungan Data Pribadi, karena wajah tergolong data biometrik, dan pemerintah melalui Kementerian Komunikasi dan Informatika telah merespons lewat Surat Edaran Nomor 9 Tahun 2023 tentang Etika Kecerdasan Artifisial yang menegaskan bahwa pemanfaatan AI harus mempertimbangkan prinsip etis, kehati-hatian, dan keselamatan.

Contoh lain adalah pelanggaran hak cipta oleh AI generatif seperti DALL·E, Midjourney, dan Stable Diffusion yang dapat meniru gaya visual karya orang lain, misalnya gaya khas animasi Studio Ghibli, karena dilatih menggunakan data karya seni yang diambil dari internet tanpa izin penciptanya. Persoalan hukum yang muncul mencakup apakah hasil karya AI dapat dilindungi hak cipta dan apakah proses pelatihan model dengan karya orang lain sudah melanggar hak cipta itu sendiri; di Indonesia sendiri, berdasarkan Undang-Undang Hak Cipta Nomor 28 Tahun 2014, hak cipta hanya diberikan kepada karya ciptaan manusia sehingga karya yang sepenuhnya dihasilkan AI secara hukum tidak memperoleh perlindungan hak cipta.

Di ranah profesional, penggunaan teknologi digital dan AI di sektor kesehatan juga rawan disalahgunakan, misalnya kasus tenaga kesehatan yang melakukan siaran langsung saat mendampingi proses persalinan pasien tanpa persetujuan yang memadai. #2. Dampak Energi dan Lingkungan dari Pemanfaatan Kecerdasan Buatan serta Cara Mengatasinya

Dampak lingkungan AI paling sering dikaitkan dengan konsumsi listrik pusat data (data center) yang menjalankan pelatihan dan operasional model AI, termasuk kebutuhan pendinginan server yang sangat besar. Sistem pendinginan saja dapat menyumbang hingga 40% dari total konsumsi energi sebuah data center, sehingga efisiensi pada aspek ini menjadi krusial untuk menekan dampak lingkungan sekaligus biaya operasional. Riset memperkirakan beban kerja AI akan menyumbang lebih dari separuh konsumsi energi data center global pada 2028 akibat pemrosesan model-model besar, dan selain listrik, operasional data center juga menghabiskan banyak air untuk pendinginan serta menghasilkan limbah elektronik yang menjadi tantangan tersendiri. Secara global, sektor TIK diperkirakan menyumbang sekitar 1,8–3,9% emisi gas rumah kaca, dan pertumbuhan AI generatif berpotensi memperbesar kontribusi tersebut secara signifikan, sebuah kondisi yang semakin kritis di Indonesia mengingat bauran energi nasional masih didominasi bahan bakar fosil.

Yang menarik, penelitian terbaru menunjukkan bahwa dampak iklim AI tidak berhenti pada konsumsi listrik data center saja. Studi yang dipublikasikan dalam jurnal NPJ Climate Action menemukan bahwa AI juga dimanfaatkan industri bahan bakar fosil untuk meningkatkan produksi minyak dan gas, sehingga secara tidak langsung mendorong aktivitas industri dengan emisi karbon tinggi; para peneliti bahkan menilai dampak ini berpotensi lebih besar dibanding manfaat AI dalam mendukung transisi energi bersih, sehingga mendorong perlunya regulasi dan pengawasan yang lebih kuat.

Dari sisi solusi, beberapa upaya mitigasi mulai dikembangkan. AI sendiri kini dipakai untuk memantau dan menyesuaikan konsumsi energi data center secara real-time, membaca pola suhu, beban kerja, dan kondisi perangkat untuk mengoptimalkan pendinginan dan penggunaan listrik, sehingga meningkatkan efisiensi dan mengurangi dampak lingkungan; di sisi kebijakan, pemerintah Indonesia juga mulai mendorong penyediaan energi hijau melalui berbagai regulasi. Penerapan standar manajemen lingkungan seperti ISO 14001 juga mulai diadopsi operator data center untuk mengendalikan dampak lingkungan sekaligus mendukung operasional AI yang lebih berkelanjutan. Kerangka green computing yang mencakup dimensi green use, green design, green disposal, dan green manufacturing turut diusulkan sebagai pendekatan menyeluruh, disertai rekomendasi berupa standar green data center nasional, percepatan transisi ke energi terbarukan, dan peningkatan kesadaran pengguna terhadap jejak lingkungan dari aktivitas digital mereka.

Menurut saya pribadi, akar masalahnya bukan pada teknologi AI itu sendiri, melainkan pada kecepatan adopsinya yang jauh melampaui kesiapan infrastruktur energi bersih. Solusi jangka pendek seperti optimalisasi pendinginan dan efisiensi algoritma memang penting, tetapi sifatnya hanya menahan laju pertumbuhan dampak, bukan menghilangkannya. Solusi yang lebih mendasar adalah mempercepat transisi sumber energi data center ke energi terbarukan dan mewajibkan transparansi jejak karbon dari setiap penyedia layanan AI, sehingga pengguna dan regulator bisa menilai dampak riilnya, bukan sekadar klaim keberlanjutan yang bersifat pemasaran. Sebagai pengguna maupun calon praktisi di bidang teknologi, kesadaran untuk menggunakan AI secara proporsional—tidak menjalankan komputasi berat untuk hal yang sebenarnya bisa diselesaikan dengan cara lebih sederhana—juga menjadi bagian tanggung jawab yang sering diabaikan.
